```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset('tips')
print(type(tips))
print(tips.head())
```

```python
sns.relplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'smoker',
    size = 'size',
    style = 'smoker'
)
```

```python
# Relational Plots
#Scatter plot
sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'time'
)
```

```python
#line plots
flights = sns.load_dataset('flights')
flights.head()
```

```python
sns.lineplot(
    data = flights,
    x = 'year',
    y = 'passengers',
)
```

```python
#Bar plot
sns.barplot(
    data = tips,
    x = 'day',
    y = 'tip',
    hue = 'sex'
)
```

```python
#Box plot
sns.boxplot(
    data = tips,
    x = 'day',
    y = 'tip'
)
```

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.head()
```

```python
#histogram
sns.histplot(
    data = penguins,
    x = 'body_mass_g',
    bins = 30
)
```

```python
#Heat maps - color encoded matrix
flight_pivot = flights.pivot(index = 'month', columns = 'year', values = 'passengers')
```

```python
sns.heatmap(
    flight_pivot,
    cmap = 'coolwarm'
)
```

```python
#Best practices

fig, ax = plt.subplots()

sns.lineplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    hue = 'sex',
    ax = ax
)

ax.set_title('avg bill per day')
ax.set_xlabel('Days')
ax.set_ylabel('Total Bill Amount')

fig.tight_layout()
```
