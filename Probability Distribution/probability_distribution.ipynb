```python
#Binomial distribution
from scipy.stats import binom

n = 5
k = 3
p = 0.5

prob = binom.pmf(k, n, p)
print(prob)
```

```python
#Uniform distribution
import numpy as np
import matplotlib.pyplot as plt

#values
values = np.random.uniform(0, 10, 10_00_000)

#plot by histogram
plt.hist(values, bins = 100, density = True, alpha = 0.3)

plt.title('Continuous Uniform Distribution between 0 and 10')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.grid(True)
```

```python
#Normal distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

#parameters
mu = 70 #mean
sigma = 10 #Standard Deviation

#X-axis
x = np.linspace(30, 110, 1000)

#Normal PDF
y = norm.pdf(x, mu, sigma)

#Plot the bell curve
plt.plot(x, y, color = 'black', linewidth = 2, label = 'Normal Distribution')

#Shade 68% region
x1 = np.linspace(mu - sigma, mu + sigma, 1000)
y1 = norm.pdf(x1, mu, sigma)
plt.fill_between(x1, y1, color = 'green', alpha = 0.3, label = '68% region')

#Shade 95% region
x2 = np.linspace(mu - 2*sigma, mu + 2*sigma, 1000)
y2 = norm.pdf(x2, mu, sigma)
plt.fill_between(x2, y2, color = 'yellow', alpha = 0.2, label = '95% region')

#Shade 99.7% region
x3 = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)
y3 = norm.pdf(x3, mu, sigma)
plt.fill_between(x3, y3, color = 'red', alpha = 0.1, label = '99.7% region')

plt.title('Normal Distribution of Student Scores')
plt.xlabel('Score')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
```

```python
# Plot y = root(x)
x = np.linspace(0, 100, 100_000)
y = np.sqrt(x)

#Plot
plt.plot(x, y, label = 'y = root(x)')
plt.title('y = root(x)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
```

```python
# Plot of sinx
x = np.linspace(-2*np.pi, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y, label = 'y = sin(x)', color = 'blue')

plt.axvline(0, color = 'black', linewidth = 1.5, linestyle = '--', label = 'x = 0')
plt.axhline(0, color = 'black', linewidth = 1.5, linestyle = '--', label = 'y = 0')

plt.title('y = sin(x)')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
```
